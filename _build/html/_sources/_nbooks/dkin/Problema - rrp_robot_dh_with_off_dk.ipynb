## Ejercicio 1 - Cinemática diferencial

```python
RRP = Robot((0,pi/2,d1,q1), (0,pi/2,-d2,q2), (0,0,q3,0,"p"))
T = RRP.T
```

```python
sym_vals = {cos(q1):symbols("c_1"), sin(q1):symbols("s_1"), cos(q2):symbols("c_2"), sin(q2):symbols("s_2")}
num_vals = {
    d1: 320,
    d2: 70,
    q1: pi/2,
    q2: pi/6,
    q3: 200,
    q1p: 0.25,
    q2p: 1.75,
    q3p: 35
}
```

```python
RRP.T_i0(1).subs(sym_vals)
```

```python
RRP.T_i0(2).subs(sym_vals)
```

```python
RRP.T_i0(3).subs(sym_vals)
```

## a) Velocidad lineal

```python
ve = T[:3,3]
ve.diff(t).subs(sym_vals), ve.diff(t).subs(num_vals).evalf(4)
```

## b) Velocidades angulares

```python
RRP.w_i(1).subs(sym_vals), RRP.w_i(2).subs(sym_vals), RRP.w_i(3).subs(sym_vals)
```

```python
RRP.w_i(1).subs(num_vals), RRP.w_i(2).subs(num_vals).evalf(4), RRP.w_i(3).subs(num_vals).evalf(4)
```

## c) $\vec{v}_Q$

```python
rQ = RRP.T_i0(2) * Matrix([0,35,0,1])
vQ = rQ.diff(t)
rQ, vQ, vQ.subs(num_vals)
```

d) $J$

```python
RRP.J.subs(sym_vals), RRP.J.subs(num_vals).evalf(4)
```
