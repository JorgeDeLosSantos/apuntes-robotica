```python
from sympy import *
init_printing()
```

```python
rho,t,R,m = symbols("t,rho,R,m", positive=True)
x,y = symbols("x,y")
```

## Momento de inercia $ I_{xx} $

$$ dm = \frac{m}{\pi R^2} \, dx \, dy $$
$$ r_x = y $$

$$ 
I_{xx} = \frac{m}{\pi R^2} \int_{-R}^{R} \int_{-\sqrt{R^2 - x^2}}^{\sqrt{R^2 - x^2}} y^2 \, dy\,dx
$$

```python
Ixx_1 = integrate(y**2, (y, -sqrt(R**2-x**2), sqrt(R**2-x**2)))
Ixx = (m/(pi*R**2)) * integrate( Ixx_1, (x, -R, R) )
Ixx
```

## Momento de inercia $ I_{yy} $

$$ dm = \frac{m}{\pi R^2} \, dx \, dy $$

$$ 
I_{yy} = \frac{m}{\pi R^2} \int_{-R}^{R} \int_{-\sqrt{R^2 - x^2}}^{\sqrt{R^2 - x^2}} x^2 \, dy\,dx
$$

```python
Iyy_1 = integrate(x**2, (y, -sqrt(R**2-x**2), sqrt(R**2-x**2)))
Iyy = (m/(pi*R**2)) * integrate( Iyy_1, (x, -R, R) )
Iyy
```

## Momento de inercia $I_{zz}$

$$ dm = \frac{m}{\pi R^2} \, dx \, dy $$
$$ r_z = x^2 + y^2 $$

$$ 
I_{zz} = \frac{m}{\pi R^2} \int_{-R}^{R} \int_{-\sqrt{R^2 - x^2}}^{\sqrt{R^2 - x^2}} y^2 \, dy\,dx
$$

```python
Izz_1 = integrate((x**2+y**2), (y, -sqrt(R**2-x**2), sqrt(R**2-x**2)))
Izz = (m/(pi*R**2)) * integrate( Izz_1, (x, -R, R) )
Izz
```

## Producto de inercia $ I_{xy} $

$$ 
I_{xy} = - \int_m xy \,dm = - \frac{m}{\pi R^2} \int_{-R}^{R} \int_{-\sqrt{R^2 - x^2}}^{\sqrt{R^2 - x^2}} xy \, dy\,dx
$$

```python
Ixy_1 = integrate(x*y, (y, -sqrt(R**2-x**2), sqrt(R**2-x**2)))
Ixy = (m/(pi*R**2)) * integrate( Ixy_1, (x, -R, R) )
Ixy
```

Condiciones de simetría, implica: $ I_{xy} = I_{xz} = I_{yz} = 0 $
