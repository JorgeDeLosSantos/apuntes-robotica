```python
from sympy import *
init_printing()
```

```python
def dh(a,alpha,d,theta):
    ct = cos(theta)
    st = sin(theta)
    ca = cos(alpha)
    sa = sin(alpha)
    H = Matrix([[ct, -st*ca, st*sa, a*ct],
                [st, ct*ca, -ct*sa, a*st],
                [0,    sa,    ca,    d],
                [0,0,0,1]])
    return H
```

```python
th = symbols("theta")
l = symbols("l")
dh(l,0,0,th)
```
