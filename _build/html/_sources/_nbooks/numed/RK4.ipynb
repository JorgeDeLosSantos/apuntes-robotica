Resolver:

$$ \frac{dy}{dx} = 3t+2 $$

Sujeta a:

$$ y(0) = 5 $$

```python
def f(t,y):
    return np.array([3*t + 2])

def ye(t):
    return 3*t**2/2 + 2*t + 5
```

```python
tsim = [0, 0.5]
y0 = [5]
h = 0.1

TE,YE = metodo_euler(f, tsim, y0, h)
TRK4,YRK4 = metodo_rk4(f, tsim, y0, h, latex=True)
```

```python
plt.plot(TE, YE, "-o", label="Aproximación: Euler")
plt.plot(TRK4, YRK4, "-o", label="Aproximación: RK4")
plt.plot(TE, ye(np.array(TE)), "k--" ,label="Solución real", alpha=0.6)
plt.xlabel("t")
plt.ylabel("y")
plt.grid(ls="--")
plt.legend()
plt.savefig(img_path+"rk4_numed_example_01.svg")
```
