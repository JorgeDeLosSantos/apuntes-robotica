```python
from sympy import *
from sympy.matrices import Matrix
from rkd.didactic import rotx,roty,rotz,theta
```

```python
alpha = pi/6
beta = 3*pi/4
Rz = Matrix([[cos(alpha), -sin(alpha), 0], [sin(alpha), cos(alpha), 0], [0,0,1]])
Rx = Matrix([[1,0,0], [0, cos(beta), -sin(beta)], [0, sin(beta), cos(beta)]])
```

```python
RBA = Rz*Rx
RAB = RBA.inv()

print(RBA.evalf())
print(RAB.evalf())
```
