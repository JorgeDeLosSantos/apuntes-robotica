```python
x,y,z = symbols("x,y,z")
a,b,c,h = symbols("a,b,c,h")
rho = symbols("rho")
```

```python
I1 = integrate(1, (x,0,a))
I2 = integrate(I1, (y,0,b/c*z))
I3 = integrate(I2, (z,0,c))
idV = I3
I3
```

```python
I1 = integrate(1, (y,0,b/c*z))
I2 = integrate(I1, (x,0,a))
I3 = integrate(I2, (z,0,c))
I3
```

Para:
$$\int_V \tilde{x} \, dV$$

```python
I1 = integrate(x, (x,0,a))
I2 = integrate(I1, (y,0,b/c*z))
I3 = integrate(I2, (z,0,c))
ixtdV = I3
I3
```

Para:
$$\int_V \tilde{y} \, dV$$

```python
I1 = integrate(y, (x,0,a))
I2 = integrate(I1, (y,0,b/c*z))
I3 = integrate(I2, (z,0,c))
iytdV = I3
I3
```

Para:
$$\int_V \tilde{z} \, dV$$

```python
I1 = integrate(z, (x,0,a))
I2 = integrate(I1, (y,0,b/c*z))
I3 = integrate(I2, (z,0,c))
iztdV = I3
I3
```

Centro de masa

```python
xcm = ixtdV / idV
ycm = iytdV / idV
zcm = iztdV / idV
xcm, ycm, zcm
```

## Utilizando un elemento diferencial en forma de placa rectangular

$$ \int_V \, dV = \frac{ab}{c} \int_0^c z\,dz  $$

```python
integrate(a*b*z/c, (z,0,c))
```

```python
integrate( (a**2 * b) / (2*c) *z, (z,0,c) )
```

```python
integrate( (a*b**2) / (2*c**2) * z**2, (z,0,c) )
```

```python
integrate( (a*b) / (c) * z**2, (z,0,c) )
```
