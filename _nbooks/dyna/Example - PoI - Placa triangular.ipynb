```python
from sympy import *
init_printing()
```

```python
x,y = symbols("x,y")
b,h = symbols("b,h")
rho,t = symbols("rho,t")
m = symbols("m")
mass_sust = {rho*t*b*h/2: m} # masa total de la placa rectangular
```

```python
I1 = integrate(x*y, (y,0,-h/b*x + h))
I2 = integrate(I1, (x,0,b))
(-rho*t*I2).subs(mass_sust)
```

```python
I1 = integrate(x*y, (y,-h/3,-h/b*x + h/3))
I2 = integrate(I1, (x,-b/3,2*b/3))
(-rho*t*I2).subs(mass_sust)
```
