```python
from moro import *
from sympy import sqrt,atan2,acos,deg,rad
```

Cinemática inversa - Manipulador RRP with offset

```python
d2 = 120
RRP = Robot((0,pi/2,0,q1), (0,pi/2,-d2,q2), (0,0,q3,0))
```

```python
T = RRP.T
T
```

Resolviendo con el método analítico:

$$ q_3 c_1 s_2 - 120 s_1 = P_x $$
$$ q_3 s_1 s_2 + 120 c_1 = P_y $$
$$ -q_3 c_2 = P_z $$

```python
Px,Py,Pz = 200,250,350
```

```python
Q3 = sqrt(Px**2 + Py**2 + Pz**2 - d2**2)
Q2 = atan2(sqrt(Px**2 + Py**2 - d2**2), -Pz)
Q1 = atan2(Py*Q3*sin(Q2) - Px*d2, Px*Q3*sin(Q2) + Py*d2)

solu=f"""
q1 = {deg(Q1).evalf(4)}°
q2 = {deg(Q2).evalf(4)}°
q3 = {Q3.evalf(4)} mm"""


print(solu)
```

```python
T.subs({q1:Q1, q2:Q2, q3:Q3}).evalf(4)
```
