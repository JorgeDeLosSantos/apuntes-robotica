```python
from sympy import *
init_printing()
```

```python
x,y = symbols("x,y")
```

Calculamos:

$$ \int_A dA  =  \int_0^{0.2} \int_0^{0.5x+0.1} dy\,dx $$

```python
I1_dA = integrate(1, (y,0,0.5*x + 0.1))
I2_dA = integrate(I1_dA, (x,0,0.2))

I2_dA
```

Calculamos: 

$$ \int_A \tilde{x}\, dA = \int_0^{0.2} \int_0^{0.5x+0.1} x\,dy\,dx $$ 

```python
I1_xdA = integrate(x, (y,0,0.5*x + 0.1))
I2_xdA = integrate(I1_xdA, (x,0,0.2))

I2_xdA
```

Calculamos: 

$$ \int_A \tilde{y}\, dA = \int_0^{0.2} \int_0^{0.5x+0.1} y\,dy\,dx $$ 

```python
I1_ydA = integrate(y, (y,0,0.5*x + 0.1))
I2_ydA = integrate(I1_ydA, (x,0,0.2))

I1_ydA.expand(), I2_ydA
```

Entonces:

$$ 
\bar{x} = \frac{ \int_A \tilde{x}\,dA }{\int_A dA} \qquad 
\bar{y} = \frac{ \int_A \tilde{y}\,dA }{\int_A dA}
$$

```python
x_cm = I2_xdA / I2_dA
y_cm = I2_ydA / I2_dA
x_cm, y_cm
```

## Elemento rectangular

```python
I_dA = integrate(0.5*x+0.1, (x,0,0.2))
I_xdA = integrate(0.5*x**2 + 0.1*x, (x,0,0.2))
I_ydA = integrate(0.125*x**2 + 0.05*x + 0.005, (x,0,0.2))
x_cm = I_xdA / I_dA
y_cm = I_ydA / I_dA

I_dA, I_xdA, I_ydA, x_cm, y_cm
```
