```python
from moro import *
from sympy import sqrt, expand, nsolve
```

```python
RRR = Robot((0,pi/2,d1,q1), (l2,0,0,q2), (l3,0,0,q3))
```

```python
Px,Py,Pz = symbols("P_x, P_y, P_z")
```

```python
ex,ey,ez = RRR.T[:3,3]
ex,ey,ez
```

```python
ey/ex
```

```python
simplify( ex**2 + ey**2 + (ez-d1)**2 )
```

```python
simplify( (ex**2 + ey**2)/(ez-d1)**2 )
```

```python
simplify( ex**2 + ey**2 )
```

```python
ec1 = ex - Px
ec2 = ey - Py
ec3 = ez - Pz
num_vals = {d1:250, l2:250, l3:250, Px:100, Py:200, Pz:300}
solu = nsolve([ec1.subs(num_vals), ec2.subs(num_vals), ec3.subs(num_vals)], [q1,q2,q3], (0,0.1,0.1))
solu
```

```python
RRR.T.subs(num_vals).subs({q1:solu[0], q2:solu[1], q3:solu[2]})
```
