La obtención del modelo dinámico de un manipulador juega un rol importante para la simulación del movimiento, el análisis de la estructura del manipulador y el diseño de los algoritmos de control. La simulación del movimiento de un manipulador permite probar estrategias de control sin necesidad de disponer físicamente del manipulador. El cálculo de las fuerzas y torques requeridos para la ejecución de los movimientos del robot proporciona información valiosa para el diseño de las juntas, transmisiones y actuadores.

El modelo dinámico de un robot consiste en una ecuación diferencial (ordinaria) vectorial en las posiciones, ya sean articulares $ \vec{q} $ o cartesianas $ \vec{x} $, generalmente de segundo orden, pudiéndose expresar como:

$$
f(\vec{q}, \vec{\dot{q}}, \vec{\ddot{q}}, \tauvec ) = \vec{0}
$$ (joint_dyn_model)
$$ 
f(\vec{x}, \vec{\dot{x}}, \vec{\ddot{x}}, \tauvec ) = \vec{0}
$$ (cartesian_dyn_model)

En las expresiones anteriores el vector $ \tauvec $ corresponde a las fuerzas y torques aplicados en las juntas por los actuadores. La ecuación {eq}`joint_dyn_model` corresponde al *modelo dinámico articular* y la ecuación {eq}`cartesian_dyn_model` al *modelo dinámico cartesiano*.

## Dinámica de cuerpo rígidos

### Ecuaciones de Newton-Euler.

La fuerza resultante que actúa en un cuerpo rígido es igual a la tasa de cambio en el tiempo de su cantidad de movimiento lineal con respecto a un sistema de referencia inercial:

$$
\Sigma \vec{F} = \dot{ \vec{P} } = m \dot{\vec{v}}_G
$$ (newton_linear_momentum)

Donde $\vec{v}_G$ es la velocidad del centro de masa del cuerpo rígido, $\Sigma \vec{F}$ es la fuerza resultante que actúa sobre el cuerpo rígido y $\vec{P}$ es la cantidad de movimiento lineal con respecto al sistema de referencia inercial. Dado que $ \vec{a}_G = \dot{\vec{v}}_G  $, la ecuación anterior se puede escribir como

$$ \Sigma \vec{F} = m \vec{a}_G $$

De la ecuación {eq}`newton_linear_momentum` se desprende un principio o teorema de conservación de la cantidad de movimiento:

> *Si la fuerza resultante en un sólido es cero, entonces $ \dot{\vec{P}} = 0 $, lo cual implica que la cantidad de movimiento lineal del sólido se conserva, es decir, permanece constante.*

El momento resultante ($\Sigma \vec{M}_O$) de las fuerzas externas que actúan en un cuerpo rígido con respecto a un punto fijo $O$ en un sistema de referencia inercial, es igual a la tasa de cambio en el tiempo de la cantidad de movimiento angular alrededor de $O$:

$$
\Sigma \vec{M}_O = \dot{\vec{H}}_O
$$ (euler_angular_momentum)

donde $\vec{H}_O$ es la cantidad de movimiento angular del cuerpo con respecto al mismo punto $O$. De la ecuación {eq}`euler_angular_momentum` se desprende también un principio de conservación de la cantidad de movimiento angular:

> *Si el momento resultante en un sólido con respecto a un punto fijo $O$ es cero, entonces $\dot{\vec{H}}_O = 0 $, lo cual implica que la cantidad de movimiento angular $\vec{H}_O$ con respecto a $O$ se conserva, es decir, permanece constante.*

La ecuación {eq}`euler_angular_momentum` se puede escribir también haciendo que el punto $O$ ahora se corresponda con la ubicación del centro de masa ($G$), es decir:

$$ 
\Sigma \vec{M}_G = \dot{\vec{H}}_G
$$ (euler_angular_momentum_com)

Donde $\Sigma \vec{M}_G $ es la sumatoria de los momentos alrededor de $G$ y $\vec{H}_G$ la cantidad de movimiento angular alrededor del centro de masa.


Consideremos un cuerpo rígido modelado como un sistema de $N$ partículas como el que se muestra en la figura. Las ecuaciones {eq}`newton_linear_momentum` y {eq}`eq:euler_angular_momentum` son aplicables para este modelo. Si las $N$ partículas son newtonianas, la cantidad de movimiento lineal y la cantidad de movimiento angular están dadas por: 

$$ 
\vec{P} = \sumin m_i \vec{v}_i
$$ (linear_momentum_particles)

$$
\vec{H}_O = \sumin \vec{R}_i \times m_i \vec{v}_i
$$ (angular_momentum_particles)

donde $m_i$ es la masa, $\vec{v}_i$ la velocidad y $\vec{R}_i$ la posición, de la i-ésima partícula, descritas en el sistema inercial.

### Ecuaciones de Euler-Lagrange

Otra manera de abordar la dinámica de un cuerpo rígido (o partícula) es utilizar las ecuaciones de Euler-Lagrange. Dado un sistema mecánico de $n$ grados de libertad, sus ecuaciones de movimiento se pueden obtener como sigue:

$$ 
\LaEq{i} \quad ; \text{ Para } i=1,2,\cdots,n
$$ (euler_lagrange_rigid_body)

Donde $\La$ es una función escalar llamada *lagrangiano*, y está dada por la diferencia entre la energía cinética y potencial del cuerpo rígido, es decir:

$$
\La = \Ki - \Po
$$ (lagrangian)

A $q_i$ y $\dot{q}_i$ se les denomina *coordenada generalizada* y *velocidad generalizada*, de forma respectiva. Un conjunto de coordenadas generalizadas $ \rbr{q_1, q_2, \cdots, q_n} $ es aquel que describe de manera completa la posición de un sistema de $n$ grados de libertad. Las velocidades generalizadas son las derivadas de las coordenadas generalizadas $ \rbr{\dot{q}_1, \dot{q}_2, \cdots, \dot{q}_n} $.

La experiencia demuestra que dadas simultáneamente las coordenadas y velocidades generalizadas se determina completamente el estado del sistema y permite, en principio, predecir su movimiento futuro. Las relaciones entre las aceleraciones, las velocidades y las coordenadas se llaman *ecuaciones de movimiento*. \cite{landau}

El término $\tau_i$ corresponde a las fuerzas generalizadas aplicadas sobre el cuerpo rígido o sistema de cuerpos rígidos.


### Energía cinética de un cuerpo rígido
$\newcommand{\W}{\boldsymbol{\omega}} \newcommand{\colvec}[3]{\begin{bmatrix}#1\\#2\\#3\end{bmatrix}} \newcommand{\w}[1]{\omega_{#1}} $
En la Figura {numref}`lagrangian_rigid_body` se muestra un cuerpo rígido, el cual está rotando con una velocidad angular $\W$ y donde la posición de la i-ésima partícula de masa $m_i$ con respecto al sistema de referencia $Gx'y'z'$ está dada por el vector $\vec{r}_{i/G}$. El sistema de referencia $Gx'y'z'$ está adherido al sólido y el sistema $Oxyz$ es un sistema de referencia inercial.

```{figure} img/dyna/lagrangian_rigid_body.svg
:name: lagrangian_rigid_body
:align: center
:width: 60%
Un cuerpo rígido en rotación.
```

La energía cinética de un cuerpo rígido es la suma de las energías cinéticas de cada una de las partículas que lo conforman:

$$
\Ki = \sumin \frac{1}{2} m_i \vec{v}_i \cdot \vec{v}_i
$$ (kinetic_energy_rigid_body)

Para el sólido de la Figura {numref}`lagrangian_rigid_body` la velocidad $\vec{v}_i$ de la partícula $i$ se puede expresar en el sistema inercial como:

$$ \vec{v}_i = \vec{v}_G + \W \times r_{i/G} $$

Sustituyendo la velocidad en la ecuación {eq}`kinetic_energy_rigid_body` se tiene:

$$ 
\begin{matrix}
\displaystyle
\Ki = \sumin \, \frac{1}{2} m_i \rbr{\vec{v}_G + \W \times \vec{r}_{i/G}} \cdot \rbr{\vec{v}_G + \W \times \vec{r}_{i/G}} \\
\displaystyle
= \frac{1}{2} \rbr{\vec{v}_G \cdot \vec{v}_G } \sumin m_i + 
\vec{v}_G \cdot \sumin m_i \rbr{ \W \times \vec{r}_{i/G} } + 
\frac{1}{2} \sumin m_i \, \rbr{ \W \times \vec{r}_{i/G} } \cdot \rbr{ \W \times \vec{r}_{i/G} }
\end{matrix}
$$ (kinetic_energy_com)

Dado que la sumatoria de las masas de cada una de las partículas corresponde a la masa total del sólido, entonces el primer término de esta ecuación se puede expresar como:

$$ 
\frac{1}{2} \rbr{\vec{v}_G \cdot \vec{v}_G } \sumin m_i = \frac{1}{2} m \rbr{\vec{v}_G \cdot \vec{v}_G } 
$$

Para el segundo término:

$$ 
\vec{v}_G \cdot \sumin m_i \, \rbr{ \W \times \vec{r}_{i/G} } = \vec{v}_G \cdot \rbr{ \W \times \sumin m_i \, \vec{r}_{i/G} }
$$

Por la definición de centro de masa, es sencillo verificar que el término:

$$ \sumin m_i \vec{r}_{i/G} = 0 $$

Lo cual implica que el segundo término de {eq}`kinetic_energy_com` se hace cero.

Para el tercer término, podemos aplicar la siguiente identidad vectorial:

$$ \vec{A} \cdot \rbr{\vec{B} \times \vec{C}} = \vec{B} \cdot \rbr{ \vec{C} \times \vec{A}} $$

Así pues:

$$ 
\frac{1}{2} \sumin m_i \, \rbr{ \W \times \vec{r}_{i/G} } \cdot \rbr{ \W \times \vec{r}_{i/G} } = 
\frac{1}{2} \W \cdot \sumin m_i \vec{r}_{i/G} \times \rbr{ \W \times \vec{r}_{i/G} }
$$

Este producto vectorial lo podemos desarrollar, para esto vamos a considerar que las componetes de los vectores $\W$ y $\vec{r}_{i/G}$ son:

$$ 
\W = \colvec{\omega_x}{\omega_y}{\omega_z}  \, ;  \quad
\vec{r}_{i/G} = \colvec{x_i}{y_i}{z_i}
$$

Entonces:

$$ 
\W \times \vec{r}_{i/G} = \colvec{\w{x}}{\w{y}}{\w{z}} \times \colvec{x_i}{y_i}{z_i} = 
\colvec{\w{y} z_i - \w{z} y_i}{-\w{x} z_i + \w{z} x_i}{\w{x} y_i - \w{y} x_i}
$$

$$
m_i \vec{r}_{i/G} \times \rbr{\W \times \vec{r}_{i/G} } = 
\colvec{m_i x_i}{m_i y_i}{m_i z_i} \times
\colvec{\w{y} z_i - \w{z} y_i}{-\w{x} z_i + \w{z} x_i}{\w{x} y_i - \w{y} x_i} = 
$$
$$ 
= \colvec{m_i y_i \rbr{\w{x} y_i - \w{y} x_i} - m_i z_i \rbr{ -\w{x} z_i + \w{z} x_i } }
{ -m_i x_i \rbr{ \w{x} y_i - \w{y} x_i } + m_i z_i \rbr{ \w{y} z_i - \w{z} y_i } }
{m_i x_i \rbr{ -\w{x} z_i + \w{z} x_i } - m_i y_i \rbr{ \w{y} z_i - \w{z} y_i } }
$$
$$ 
= \colvec{m_i y_i^2 \w{x} - m_i x_i y_i \w{y} + m_i z_i^2 \w{x} - m_i x_i z_i \w{z}}
{ -m_i y_i x_i \w{x} + m_i x_i^2 \w{y} + m_i z_i^2 \w{y} - m_i y_i z_i \w{z} }
{ -m_i z_i x_i \w{x} + m_i x_i^2 \w{z} - m_i z_i y_i \w{y} + m_i y_i^2 \w{z} }
$$

Factorizando las velocidades angulares, cada una de las componentes del producto vectorial resulta en:

$$
\colvec{
\sbr{ m_i \rbr{y_i^2 + z_i^2} } \w{x} +
\sbr{ - m_i x_i y_i } \w{y} + 
\sbr{ - m_i x_i z_i } \w{z} }
{
\sbr{ - m_i y_i x_i } \w{x} +
\sbr{  m_i \rbr{x_i^2 + z_i^2} } \w{y} + 
\sbr{ - m_i y_i z_i } \w{z}
}{
\sbr{ - m_i z_i x_i } \w{x} +
\sbr{ - m_i z_i y_i } \w{y} + 
\sbr{ m_i \rbr{x_i^2 + y_i^2 } } \w{z} }
$$

De tal manera que, en forma más compacta:

$$
 \sumin m_i \vec{r}_{i/G} \times \rbr{ \W \times \vec{r}_{i/G} } = 
\colvec{
I_{xx} \w{x} + I_{xy} \w{y} + I_{xz} \w{z}  }
{ I_{yx} \w{x} + I_{yy} \w{y} + I_{yz} \w{z} }
{ I_{zx} \w{x} + I_{zy} \w{z} + I_{zz} \w{z} }
$$ (angular_momentum_three_equations)

donde:

$$
\begin{matrix}
\displaystyle
I_{xx} = \sumin m_i \rbr{y_i^2 + z_i^2} \\
\displaystyle
I_{yy} = \sumin m_i \rbr{x_i^2 + z_i^2} \\
\displaystyle
I_{zz} = \sumin m_i \rbr{x_i^2 + y_i^2}
\end{matrix}
$$ (moments_of_inertia)

$$
\begin{matrix}
I_{xy} = I_{yx} = - \sumin m_i x_i y_i \\
I_{xz} = I_{zx} = - \sumin m_i x_i z_i \\
I_{yz} = I_{zy} = - \sumin m_i y_i z_i 
\end{matrix}
$$ (products_of_inertia)

Las cantidades dadas en la ecuación {eq}`moments_of_inertia` se denominan *momentos de inercia* y las de la ecuación {eq}`products_of_inertia` corresponden a los *productos de inercia*. 

El vector de la ecuación {eq}`angular_momentum_three_equations` se puede escribir en forma de una multiplicación matricial como:

$$
I_G \W
$$ (angular_momentum_matricial_form)

donde $I_G$ es una matriz dada por:

$$
I_G = \begin{bmatrix}
I_{xx} & I_{xy} & I_{xz} \\ 
I_{yx} & I_{yy} & I_{yz} \\
I_{zx} & I_{zy} & I_{zz}
\end{bmatrix}
$$ (inertia_tensor)

a esta matriz se le denomina *matriz de inercia* o más comunmente *tensor de inercia*. En este caso el tensor de inercia describe la manera en que la masa está distribuida con respecto al sistema de referencia $Gx'y'z'$ adherido al sólido, así pues aún cuando el cuerpo se traslade o rote, el tensor de inercia permanece invariable. En secciones posteriores nos centraremos en cómo determinar cada uno de los componentes del tensor de inercia. 

Considerando todo lo anterior, el tercer término de {eq}`kinetic_energy_com` se puede escribir como:

$$ 
\frac{1}{2} \W \cdot \sumin m_i \vec{r}_{i/G} \times \rbr{ \W \times \vec{r}_{i/G} } = 
\frac{1}{2} \W \cdot \rbr{ I \W }
$$

Dado que el producto punto de dos vectores se puede expresar como una multiplicación matricial:

$$ \vec{u} \cdot \vec{v} = \vec{u}^T \vec{v} $$

Entonces, la energía cinética total del cuerpo rígido se puede escribir de la siguiente manera:

$$
\Ki = \frac{1}{2} m \vec{v}_G^T \vec{v}_G + \frac{1}{2} \W^T I_G \W
$$ (kinetic_energy_rigid_body_final)


#### Energía potencial de un cuerpo rígido

La energía potencial gravitatoria de un cuerpo rígido es una cantidad que depende de la masa, la gravedad y la posición del centro de masa. Se puede expresar como:

$$
\Po = - m \vec{g}^T \vec{r}_G
$$

Donde $m$ es la masa total del cuerpo rígido, $\vec{g}$ es un vector que define la aceleración de la gravedad en el sistema inercial y $\vec{r}_G$ es un vector que describe la posición del centro de masa del cuerpo rígido en el sistema inercial.	

### Centro de masa

El centro de masa es un punto hipotético de un sistema de partículas o de un sólido, tal que dicho sistema se comporta como si toda su masa estuviera concentrada en ese único punto. Es importante destacar que el centro de masa no siempre coincide con un punto material físico en el objeto. El centro de masa es el único punto en el centro de una distribución de masa en el espacio que tiene la propiedad de que los vectores de posición ponderados relativos a este punto suman cero. 

Para un sistema de $n$ partículas, cada una de masa $m_i$ y con una posición dada por $\vec{r}_i$, las coordenadas $\vec{R}$ del centro de masa deben satisfacer la condición:

$$
\displaystyle \sum_{i=1}^n m_i \left( \vec{r}_i - \vec{R}\right) = \vec{0}
$$

Resolviendo para $\vec{R}$ se tiene que:

$$
\vec{R} = \frac{\displaystyle \sum_{i=1}^n m_i \vec{r}_i }{\displaystyle \sum_{i=1}^n m_i}
$$

O bien, expresado en términos de las componentes escalares:

$$
\bar{x} = \frac{\displaystyle \sum x_i m_i }{\displaystyle \sum m_i } \, ; \quad
\bar{y} = \frac{\displaystyle \sum y_i m_i }{\displaystyle \sum m_i } \, ; \quad
\bar{z} = \frac{\displaystyle \sum z_i m_i }{\displaystyle \sum m_i } \, ; \quad
$$ (center_of_mass_particles)

Donde $x_i, y_i, z_i$ son las coordenadas de la ubicación de la partícula $i$ cuya masa es $m_i$. 

---
**Ejemplo.** *Calcule el centro de masa del sistema de partículas que se muestran en la Figura. Cada partícula tiene una masa de 0.25 kg. Las unidades de longitud de la rejilla son metros.*

```{figure} img/dyna/examples/example_particles_center_of_mass.svg
:name: example_particles_center_of_mass
:width: 50%
:align: center
.
```

**Solución:**

De acuerdo con lo que sabemos todas las masas son iguales a 0.25 kg, es decir:

$$ m_1 = m_2 = m_3 = m_4 = 0.25 \text{ kg} $$

Las ubicaciones de las partículas son:

$$ 
\rbr{x_1, y_1} = \rbr{2,0} \text{ m } ; \quad 
\rbr{x_2, y_2} = \rbr{2,2} \text{ m } ; \quad 
$$ 
$$
\rbr{x_3, y_3} = \rbr{-2,2} \text{ m } ; \quad 
\rbr{x_4, y_4} = \rbr{-2,0} \text{ m }
$$

$$ 
\bar{x} = \frac{\sum x_i m_i}{\sum m_i} = \frac{ (2)(0.25)+(2)(0.25)+(-2)(0.25)+(-2)(0.25) }{0.25+0.25+0.25+0.25} = \frac{0}{1} 
$$

$$ \bar{x} = 0  $$

$$
\bar{y} = \frac{\sum x_i m_i}{\sum m_i} = \frac{ (0)(0.25)+(2)(0.25)+(2)(0.25)+(0)(0.25) }{0.25+0.25+0.25+0.25} = \frac{1}{1}
$$

$$ \bar{y} = 1 \text{ m} $$

Las ecuaciones {eq}`center_of_mass_particles` se pueden generalizar para un cuerpo rígido, bajo la asunción de que un cuerpo rígido se puede considerar como un conjunto infinito de partículas cuyas distancias son invariables entre sí, así la sumatoria la podemos reemplazar por una integral, es decir:

$$
\bar{x} = \frac{\displaystyle \int_m \tilde{x} \, dm }{\displaystyle \int_m dm } \, ; \quad
\bar{y} = \frac{\displaystyle \int_m \tilde{y} \, dm }{\displaystyle \int_m dm } \, ; \quad
\bar{z} = \frac{\displaystyle \int_m \tilde{z} \, dm }{\displaystyle \int_m dm } \, ; \quad
$$ (center_of_mass_rigid_body)

En las ecuaciones anteriores los términos $\tilde{x}, \tilde{y}, \tilde{z}$ denotan la ubicación del centro de masa del elemento diferencial utilizado para la integración. El diferencial de masa $dm$ se puede expresar como:

$$ dm = \rho \, dV $$

En el caso general, la densidad podría no ser constante y depender de la ubicación, es decir $\rho = \rho(x,y,z)$. Para la mayoría de situaciones en ingeniería, los componentes mecánicos están fabricados con materiales homogéneos (densidad constante), lo cual implica que el centro de masa coincidará con el centroide del volumen del sólido, de tal manera que las coordenadas del centro de masa se pueden determinar mediante:

$$
\bar{x} = \frac{\displaystyle \int_V \tilde{x} \, dV }{\displaystyle \int_V dV } \, ; \quad
\bar{y} = \frac{\displaystyle \int_V \tilde{y} \, dV }{\displaystyle \int_V dV } \, ; \quad
\bar{z} = \frac{\displaystyle \int_V \tilde{z} \, dV }{\displaystyle \int_V dV } \, ; \quad
$$ (center_of_volume)

Para cuerpos rígidos con forma de placas delgadas, de espesor uniforme y homogéneas, el centro de masa coincidirá con el centroide del área de la placa, así, las coordenadas del centro de masa se pueden determinar mediante:

$$ 
\bar{x} = \frac{\displaystyle \int_A \tilde{x} \, dA }{\displaystyle \int_A dA } \, ; \quad
\bar{y} = \frac{\displaystyle \int_A \tilde{y} \, dA }{\displaystyle \int_A dA } \, ; \quad
\bar{z} = \frac{\displaystyle \int_A \tilde{z} \, dA }{\displaystyle \int_A dA } \, ; \quad
$$ (center_of_area)

```{figure} img/dyna/examples/example_triangular_plate.svg
.
```

**Solución:**

Al tratarse de una placa delgada y homogénea, las coordenadas del centro de masa se pueden determinar como:

$$ 
\bar{x} = \frac{\displaystyle \int_A \tilde{x} \, dA }{\displaystyle \int_A dA } \, ; \quad
\bar{y} = \frac{\displaystyle \int_A \tilde{y} \, dA }{\displaystyle \int_A dA } \, ; \quad
$$

Estas integrales de área se pueden plantear de varias maneras, todo depende de la elección del elemento diferencial. En la siguiente figura se muestran dos posibles opciones. Del lado izquierdo se utiliza un elemento diferencial cuya área es $dA = dx\,dy$, y cuyas coordenadas centroidales son $\tilde{x} = x$ y $ \tilde{y} = y $. Del lado derecho se utiliza un elemento diferencial cuya área es $dA = \frac{h}{b}x \, dx $, sus coordenadas centroidales son $ \tilde{x} = x $ y $ \tilde{y} = \frac{h}{2b}x $, observe que este es un elemento diferencial rectangular cuya base es $dx$, y siendo su altura aquella definida por la hipotenusa del triángulo, cuya función es $y(x) = \frac{h}{b}x$. Naturalmente, su coordenada centroidal $\tilde{y}$ está dada por la mitad del valor de $y(x)$.

```{figure} img/dyna/examples/example_triangular_plate_differential_elements.svg
:name: example_triangular_plate_differential_elements
:align: center
:width: 60%
.
```

Aquí podemos utilizar cualquiera de los dos elementos diferenciales. Sólo debemos tener en consideración que si se utiliza el elemento diferencial de $(a)$, entonces resultará una integral doble, que deberemos integrar en los límites definidos por el área triangular. Si utilizamos el elemento diferencial $(b)$ se tendrá una integral simple, que deberemos integrar en los límites definidos para la dirección en $x$, es decir, entre $0$ y $b$. Para ejemplificar vamos a proceder de ambas maneras. \\

**Utilizando el elemento diferencial $(a)$**

Calculando cada una de las integrales:

$$ 
\int_A dA = \dint{0}{b}{0}{hx/b}{}{dy}{dx} = \sint{0}{b}{\sbr{y}_0^{hx/b}}{dx} = 
\frac{h}{b} \sint{0}{b}{x}{dx} = \frac{h}{b}\sbr{\frac{x^2}{2}}_0^b = 
\frac{h}{b}\rbr{\frac{b^2}{2}} = \frac{bh}{2}
$$

$$
\int_A \tilde{x} \, dA = \dint{0}{b}{0}{hx/b}{x}{dy}{dx} = 
\sint{0}{b}{x\sbr{y}_0^{hx/b}}{dx} = 
\frac{h}{b} \sint{0}{b}{x^2}{dx} = 
$$ 
$$
= \frac{h}{b} \fsbr{x^3}{3}_0^b =
\frac{h}{b} \frbr{b^3}{3} = \frac{hb^2}{3}
$$

$$
\int_A \tilde{y} \, dA = \dint{0}{b}{0}{hx/b}{y}{dy}{dx} = 
\sint{0}{b}{\fsbr{y^2}{2}_0^{hx/b}}{dx} = 
\frac{h^2}{2b^2} \sint{0}{b}{x^2}{dx} = 
$$
$$
= \frac{h^2}{2b^2} \fsbr{x^3}{3}_0^b =
\frac{h^2}{2b^2} \frbr{b^3}{3} = \frac{h^2 b}{6}
$$

Sustituyendo los valores calculados:

$$ 
\bar{x} = \ffrac{hb^2}{3}{bh}{2} = \frac{2hb^2}{3bh} = \frac{2}{3} b
$$

$$ 
\bar{y} = \ffrac{h^2 b}{6}{bh}{2} = \frac{2h^2 b}{6bh} = \frac{1}{3} h
$$

$$ \rbr{\bar{x}, \bar{y}} = \rbr{\frac{2}{3} b, \frac{1}{3} h} $$

**Utilizando el elemento diferencial $(b)$}**

En este caso las integrales simples se pueden plantear de la siguiente manera:

$$ \int_A dA = \frac{h}{b} \sint{0}{b}{x}{dx} = \frac{bh}{2} $$
$$ \int_A \tilde{x}\, dA = \frac{h}{b} \sint{0}{b}{x^2}{dx} = \frac{hb^2}{3} $$
$$ \int_A \tilde{y}\, dA = \frac{h^2}{2b^2} \sint{0}{b}{x^2}{dx} = \frac{h^2 b}{6} $$

Observe que los resultados de las integrales son exactamente los mismos que obtuvimos con la manera anterior. La elección del elemento diferencial dependerá completamente de cuál le resulte más cómodo de plantear y resolver. En el caso de que se tenga una placa con densidad variable, aquí sí o sí tendrá que usar una integral doble para determinar el centro de masa.
