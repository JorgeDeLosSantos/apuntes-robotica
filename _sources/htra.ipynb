# Posición, orientación y transformaciones homogéneas

$$
\renewcommand{\vec}[1]{\boldsymbol{\mathbf{#1}}}

\newcommand{\framei}[1]{ \{#1\} }
$$

Un manipulador serial es un conjunto de eslabones rígidos unidos mediante pares cinemáticos. En el estudio de la cinemática usualmente lo que haremos es colocar sistemas de referencias adheridos a cada uno de los eslabones y posteriormente encontrar las relaciones entre estos sistemas para poder realizar una descripción de la cinemática del manipulador. Para lograr lo anterior,  necesitamos conocer los conceptos y herramientas matemáticas fundamentales para la representación de la posición y orientación de un cuerpo rígido, y eso es justamente el objetivo de este capítulo.

## Representación de la posición

En mecánica, la posición de un punto se describe en un sistema de referencia mediante vectores de posición. En la Figura {numref}`rep_pos_01` se muestran dos puntos $P$ y $Q$, cuyas posiciones están dadas por:

$$ \vec{r}_P = \colvec{6}{3}{0}  \qquad ; \qquad \vec{r}_Q = \colvec{2}{5}{0}  $$ 

Naturalmente ambos vectores están descritos en el sistema de referencia $xyz$ que se muestra, y dado que únicamente se tiene dicho sistema de referencia, no hay necesidad de especificar con respecto a qué sistema se están midiendo las coordenadas de posición representadas.


En la Figura {numref}`rep_pos_02` se muestra un punto $P$ y dos sistemas de referencia $\{0\}$ y $\framei{1}$. Donde el sistema $\framei{1}$ se obtiene mediante operaciones de  traslación a partir de $\framei{1}$. La descripción de $P$ en cada uno de los sistemas de referencia está dada por:

$$
\vec{r}_P^0 = \colvec{5}{2}{0} \qquad ; \qquad \vec{r}_P^1 = \colvec{2}{-3}{0}
$$

Es más, notará que tal como se muestra en la Figura {numref}`rep_pos_03` las posiciones descritas en ambos sistemas están relacionadas mediante:

$$
\vec{r}_P^{0} = \vec{r}_{o_1}^0 + \vec{r}_P^{1}
$$ (rep_pos)

Donde $ \vec{r}_{o_1}^0 $ describe la posición del origen de coordenadas del sistema $\framei{1}$ con respecto al sistema $\framei{0}$. Se debe tener cuidado con la ecuación {eq}`rep_pos`, puesto que únicamente se cumple cuando los dos sistemas de referencia involucrados son paralelos entre sí, en cualquier otra situación una expresión como esa no será válida.

```{figure} img/htra/rep_pos_02.svg
:name: rep_pos_02
:width: 50%
:align: center
Representación de la posición.
```

Otra opción es decribir las componentes de los ejes del sistema de referencia $ \{1\} $ en el sistema $ \{0\} $, es decir, $ \vec{x}_1^0 $ y $ \vec{y}_1^0 $. En el caso particular del ejemplo de la figura \ref{fig:rep_or_01} se puede observar que dichos vectores estarán dados por:

$$ \vec{x}_1^0 = \colvec{\cos\theta}{\sin\theta}{}  \qquad ; \qquad  \vec{y}_1^0 = \colvec{-\sin\theta}{\cos\theta}{} $$ 

Si estos vectores se acomodan de tal manera que:

$$ 
R_1^0 = \begin{bmatrix}
\vec{x}_1^0 & \vec{y}_1^0
\end{bmatrix}  = 
\begin{bmatrix}
\cos\theta & -\sin\theta \\
\sin\theta & \cos\theta
\end{bmatrix}
$$ (rotation_matrix_2x2)

Se obtiene un arreglo conocido como \textbf{matriz de rotación}, la cual presenta algunas propiedades particulares que se describirán posteriormente. La matriz $ R_1^0 $ contiene la información necesaria para describir la orientación del sistema $ \{1\} $ con respecto a $\{0\}$. 

Un resultado similar se puede obtener si la descripción del sistema $ \{1\} $ en $ \{0\} $ se hace mediante la proyección escalar de cada uno de los ejes, por ejemplo,  $ \vec{x}_1^0 $ y  $ \vec{y}_1^0 $ se pueden expresar como:

$$ 
\vec{x}_1^0 = \colvec{ \vec{x}_1 \cdot \vec{x}_0 }{ \vec{x}_1 \cdot \vec{y}_0 }{} 
= \colvec{\cos\theta}{\cos(90°-\theta)}{} = \colvec{\cos\theta}{\sin\theta}{}
$$


$$
\vec{y}_1^0 = \colvec{ \vec{y}_1 \cdot \vec{x}_0 }{ \vec{y}_1 \cdot \vec{y}_0 }{} = 
\colvec{\cos(90°+\theta)}{\cos\theta}{} = \colvec{-\sin\theta}{\cos\theta}{}
$$ 


Recordar que en el caso de vectores unitarios la proyección escalar de un vector sobre el otro resulta simplemente en el coseno del ángulo $\theta$ que forman entre ellos.

La descripción de los tres ejes coordenados se puede completar considerando que el eje $z_1$ permanece invariable, apuntando en la misma dirección que el eje $z_0$, tal como se puede observar en la Figura {ref}`basic_rot`. Así, es posible ampliar la matriz de la ecuación {eq}`rotation_matrix_2x2` considerando ahora las componentes en la dirección de $z$, es decir:

$$
R_1^0 = \begin{bmatrix}
\vec{x}_1^0 & \vec{y}_1^0 & \vec{z}_1^0
\end{bmatrix} 
= 
\begin{bmatrix}
\vec{x}_1 \cdot \vec{x}_0 & \vec{y}_1 \cdot \vec{x}_0 & \vec{z}_1 \cdot \vec{x}_0 \\ 
\vec{x}_1 \cdot \vec{y}_0 & \vec{y}_1 \cdot \vec{y}_0 & \vec{z}_1 \cdot \vec{y}_0 \\
\vec{x}_1 \cdot \vec{z}_0 & \vec{y}_1 \cdot \vec{z}_0 & \vec{z}_1 \cdot \vec{z}_0 \\
\end{bmatrix}
=
\begin{bmatrix}
\cos\theta & \cos(90°+\theta) & \cos 90° \\
\cos(90°-\theta) & \cos\theta & \cos 90° \\
\cos 90° & \cos 90° & \cos 0°
\end{bmatrix}
$$
$$
R_1^0 = 
\begin{bmatrix}
\cos\theta & -\sin\theta & 0 \\
\sin\theta & \cos\theta & 0 \\
0 & 0 & 1
\end{bmatrix}
$$
