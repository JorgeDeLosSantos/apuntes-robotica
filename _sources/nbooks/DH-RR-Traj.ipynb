```python
from sympy import *
import matplotlib.pyplot as plt
import numpy as np
init_printing(use_latex=true)
%matplotlib inline
```

```python
l1,l2 = 200, 250
t1,t2 = symbols("\\theta_1, \\theta_2")
x = l1*cos(t1) + l2*cos(t1+t2)
y = l1*sin(t1) + l2*sin(t1+t2)

for t in np.linspace(0, 2*np.pi, 20):
    T1 = t
    T2 = -t
    
    # Extremo
    Xe = x.subs({t1:T1, t2:T2})
    Ye = y.subs({t1:T1, t2:T2})
    # O1
    O1x = l1*cos(T1)
    O1y = l1*sin(T1)
    plt.plot([0,O1x,Xe],[0,O1y,Ye], ls="--", color="#aaddaa")
    plt.plot(Xe,Ye,"bo")
    plt.axis('equal')
```
