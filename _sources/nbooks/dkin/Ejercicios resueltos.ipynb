## Ejercicio 1

Para el manipulador planar RR de la Figura \ref{fig:rr_robot_dh}, calcule la velocidad de su extremo para un instante en el cual $ q_1 = 45° $ y $ q_2 = 30° $, siendo $ \dot{q}_1 = 1.5 $ rad/s y $ \dot{q}_2 = 3 $ rad/s. Considere que $ l_1 = l_2 = 280 $ mm. 

```python
RR = Robot((l1,0,0,q1), (l2,0,0,q2))
```

```python
vals = {q1:pi/4, q2:pi/6, q1p:1.5, q2p:3, l1:280, l2:280}
```

```python
J = RR.J.subs(vals).evalf(6)
qp = Matrix([1.5,3])
```

```python
( J * qp ).evalf(6)
```

Considere que $ a = 200 $ mm y $ b = 70 $ mm.

```python
a,b = symbols("a,b")
RPP = Robot((b,0,a,q1), (0,pi/2,q2,pi/2,"p"), (0,0,q3,0,"p"))
```
