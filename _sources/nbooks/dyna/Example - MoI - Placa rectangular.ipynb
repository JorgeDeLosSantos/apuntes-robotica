```python
from sympy import *
init_printing()
```

```python
x,y = symbols("x,y")
b,h = symbols("b,h")
rho,t,m = symbols("rho,t,m")
mass_sust = {rho*t*b*h: m} # masa total de la placa rectangular
```

Calculando $I_{x_1x_1}$

$$ 
I_{x_1x_1} = \int_m r_x^2 \, dm = \rho t  \int_{-\frac{b}{2}}^{\frac{b}{2}} \int_{-\frac{h}{2}}^{\frac{h}{2}} y_1^2 \, dy_1 \, dx_1 
$$

```python
I1 = integrate(y**2, (y,-h/2,h/2))
I2 = integrate(I1, (x, -b/2,b/2))

( rho*t*I2 ).subs(mass_sust)
```

$$ 
I_{y_1y_1} = \int_m r_y^2 \, dm = \rho t  \int_{-\frac{b}{2}}^{\frac{b}{2}} \int_{-\frac{h}{2}}^{\frac{h}{2}} x_1^2 \, dy_1 \, dx_1 
$$

```python
I1 = integrate(x**2, (y,-h/2,h/2))
I2 = integrate(I1, (x, -b/2,b/2))

( rho*t*I2 ).subs(mass_sust)
```

$$ 
I_{z_1z_1} = \int_m r_z^2 \, dm = \rho t  \int_{-\frac{b}{2}}^{\frac{b}{2}} \int_{-\frac{h}{2}}^{\frac{h}{2}} \left(x_1^2 + y_1^2 \right)\, dy_1 \, dx_1 
$$

```python
I1 = integrate(x**2 + y**2, (y,-h/2,h/2))
I2 = integrate(I1, (x, -b/2,b/2))

( factor( rho*t*I2 ) ).subs(mass_sust)
```
