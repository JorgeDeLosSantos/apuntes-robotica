```python
from sympy import *
init_printing()
```

```python
x,y,z = symbols("x,y,z")
a,b,c,h = symbols("a,b,c,h")
rho = symbols("rho")
```

## Placa triangular

```python
i1 = integrate(1, (y, 0, h*x/b))
i2 = integrate(i1, (x, 0, b))
i2
```

```python
i1 = integrate(1, (x, 0, b))
i2 = integrate(i1, (y, 0, y*b/h))
i2
```

## Prisma rectangular

```python
i1 = integrate(x, (x,0,a))
i2 = integrate(i1, (y,0,b))
i3 = integrate(i2, (z,0,c))
rho * i3
```

```python
i1 = integrate(y, (x,0,a))
i2 = integrate(i1, (y,0,b))
i3 = integrate(i2, (z,0,c))
rho * i3
```

```python
i1 = integrate(z, (x,0,a))
i2 = integrate(i1, (y,0,b))
i3 = integrate(i2, (z,0,c))
rho * i3
```
