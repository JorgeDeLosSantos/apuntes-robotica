```python
from moro import *
from sympy import sqrt, trigsimp
```

```python
Rz = lambda x: rotz(x)
Ry = lambda x: roty(x)
Rx = lambda x: rotx(x)
```

```python
R = Rz(alpha) * Ry(beta) * Rz(theta) * Ry(-beta) * Rz(-alpha)
R
```

```python
kx,ky,kz = symbols("k_x, k_y, k_z")
```

```python
K = sqrt(kx**2 + ky**2)
trigsimp( R.subs({cos(alpha):kx/K, sin(alpha):ky/K, cos(beta):kz, sin(beta):K}) )
```
