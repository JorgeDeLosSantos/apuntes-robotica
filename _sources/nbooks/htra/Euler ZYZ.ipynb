```python
from moro import *
```

```python
red_vals = {
    cos(phi): symbols("c\phi"),
    sin(phi): symbols("s\phi"),
    cos(theta): symbols("c\\theta"),
    sin(theta): symbols("s\\theta"),
    cos(psi): symbols("c\psi"),
    sin(psi): symbols("s\psi"),
}
```

```python
rotz(phi)
```

```python
roty(theta)
```

```python
rotz(psi)
```

```python
Rzyz = rotz(phi)*roty(theta)*rotz(psi)
Rzyz
```

```python
Rzyz.subs(red_vals)
```

```python
simplify(Rzyz.subs({theta:0}))
```

```python
simplify(Rzyz.subs({theta:pi}))
```
