```python
from moro import *
from sympy import asin,acos,atan2,deg,sin,cos,tan
```

```python
ppr = Robot((0,pi/2,d1,pi/2,"p"), (0,pi/2,d2,-pi/2,"p"), (l3,0,0,t3,"r"))
T = ppr.T
J = ppr.J
```

```python
TA = ( htmtra([300,0,150])*htmrot(pi/2,"x")*htmrot(pi/4,"z") )
TA
```

```python
TB = ( htmtra([500,0,250])*htmrot(pi/2,"x"))
TB
```

```python
# Para TA
L3 = 100
T3 = atan2(TA[2,1], TA[0,1])
D2 = TA[0,3] - L3*sin(T3)
D1 = TA[2,3] + L3*cos(T3)

deg(T3).evalf(), D1.evalf(), D2.evalf(), T.subs({d1:D1, d2:D2, t3:T3, l3:L3}).evalf(6, chop=True)
```

```python
# Para TB
L3 = 100
T3 = atan2(TB[2,1], TB[0,1])
D2 = TB[0,3] - L3*sin(T3)
D1 = TB[2,3] + L3*cos(T3)

deg(T3).evalf(), D1.evalf(), D2.evalf(),  T.subs({d1:D1, d2:D2, t3:T3, l3:L3}).evalf(6, chop=True)
```
