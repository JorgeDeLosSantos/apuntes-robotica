```python
from moro import *
```

```python
Rx = lambda x: htmrot(x, "x")
Ry = lambda x: htmrot(x, "y")
Rz = lambda x: htmrot(x, "z")
Dx = lambda x: htmtra([x,0,0])
Dy = lambda x: htmtra([0,x,0])
Dz = lambda x: htmtra([0,0,x])
```

```python
HB0 = Dx(400) * Dz(300) * Rx(pi/2) * Rz(pi/6)
HAB = Dx(-90) * Rz(-pi/2)
HA0 = HB0 * HAB

HB0 #, HA0
```

```python
l2 = 250
l3 = 200
PRR = Robot((0,pi/2,q1,0), (l2,0,0,q2), (l3,0,0,q3))
```

```python
T = PRR.T
```

```python
solB = solve(T - HB0)
```

```python
solB[0][q1].evalf(), (solB[0][q2]*180/pi).evalf(), (solB[0][q3]*180/pi).evalf()
```

```python
solB[1][q1].evalf(), (solB[1][q2]*180/pi).evalf(), (solB[1][q3]*180/pi).evalf()
```

```python
solA = solve(T - HA0)
```

```python
solA[0][q1].evalf(), (solA[0][q2]*180/pi).evalf(), (solA[0][q3]*180/pi).evalf()
```
