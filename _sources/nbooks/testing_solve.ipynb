```python
from sympy import *
from sympy.abc import x,y,z,t
from sympy.matrices import Matrix
init_printing(use_latex=true)

t1,t2,t3,t4 = symbols("\\theta_1:5")
r1,r2,r3,r4 = symbols("r_1:5")
w2,w3,w4 = symbols("\\omega_2:5")

def newton_raphson(J,b,X0,vals,eps=0.001):
    ei, k = [], 1
    Jinv = J.inv().subs(vals)
    b = b.subs(vals)
    while True:
        x = (Jinv.subs(X0)*b.subs(X0)).evalf()
        ei.append([x.norm(),X0[t3],X0[t4]])
        if x.norm()<eps: break
        for jj,ky in enumerate(X0):
            X0[ky] += (x[jj]).evalf()
        k += 1
    return X0,x,k,ei
```
