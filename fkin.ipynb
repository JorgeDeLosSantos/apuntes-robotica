# Cinemática directa

La **cinemática directa** permite determinar la posición y orientación del elemento terminal de un manipulador a partir de sus coordenadas articulares. Es uno de los problemas fundamentales de la robótica y constituye la base para estudiar posteriormente la cinemática inversa, las velocidades y la dinámica.

En el capítulo anterior introdujimos las matrices de rotación y las transformaciones homogéneas para describir sistemas de referencia. Ahora emplearemos esas herramientas para construir modelos de manipuladores seriales. Comenzaremos con la composición directa de transformaciones y después estudiaremos la convención de Denavit–Hartenberg (DH), que facilita la asignación sistemática de sistemas.

La Figura {numref}`fk_and_ik` ilustra la relación entre cinemática directa e inversa. La primera parte de las coordenadas articulares; la segunda busca configuraciones articulares que produzcan una pose deseada.


Al finalizar el capítulo, deberías poder describir la cadena cinemática de un manipulador serial, establecer sistemas de referencia apropiados, obtener un modelo de cinemática directa y comprobar que sus resultados son coherentes con la geometría del robot.

## El problema de la cinemática directa

Consideremos un manipulador serial con $n$ articulaciones de un grado de libertad. Representamos las variables articulares mediante el vector

$$
q=\begin{bmatrix}q_1&q_2&\cdots&q_n\end{bmatrix}^T.
$$

Una **articulación revoluta (R)** permite la rotación relativa entre dos eslabones alrededor de un eje; su variable $q_i$ es un ángulo. Una **articulación prismática (P)** permite la traslación relativa a lo largo de un eje; su variable $q_i$ es un desplazamiento lineal. Ambas poseen un grado de libertad, y los valores de las variables articulares determinan la **configuración** del manipulador.

También se utilizan los términos *junta revoluta* y *junta prismática* como sinónimos de articulación. En teoría de mecanismos, al describir el movimiento relativo permitido entre los eslabones, es habitual hablar de *pares cinemáticos*: *par revoluto* y *par prismático*. En adelante, utilizaremos preferentemente **articulación revoluta** y **articulación prismática**.

La cinemática directa busca la transformación homogénea que describe la pose del sistema terminal $\{n\}$ respecto del sistema de la base $\{0\}$:

$$
T_n^0(q)=
\begin{bmatrix}
R_n^0(q)&\vec r_{o_n}^{\,0}(q)\\
\vec 0^{\,T}&1
\end{bmatrix}.
$$

El bloque $R_n^0(q)$ representa la orientación del sistema terminal y el vector $\vec r_{o_n}^{\,0}(q)$ representa la posición de su origen, ambos respecto de la base. Así, una transformación homogénea describe la **pose**, no únicamente la posición.

Aunque la cinemática directa se calcula componiendo transformaciones, la expresión obtenida depende de la geometría del manipulador y de los sistemas elegidos. Un mismo robot admite diferentes descripciones matemáticas sin que cambie su comportamiento físico.

### Espacio articular y espacio operacional

El **espacio articular** reúne las configuraciones del robot, expresadas por los valores de $q_1,\ldots,q_n$. Sus coordenadas no tienen por qué compartir unidades: las articulaciones revolutas se describen mediante ángulos y las prismáticas mediante distancias.

El **espacio operacional** representa las tareas en términos de variables asociadas a la posición y orientación del elemento terminal. Si únicamente interesa posicionar un punto, pueden bastar sus coordenadas $(x,y,z)$. Si también importa la orientación, necesitamos una representación adicional, como las matrices de rotación estudiadas en el capítulo anterior.

La cinemática directa establece una correspondencia desde las configuraciones articulares hacia las poses alcanzadas:

$$
q\longmapsto T_n^0(q).
$$

No todas las poses son necesariamente alcanzables: la arquitectura, las dimensiones y los límites articulares restringen los movimientos posibles. Por otra parte, diferentes configuraciones pueden dar lugar a una misma pose; estudiaremos esta cuestión en el capítulo de cinemática inversa.

## Cadenas cinemáticas y sistemas de referencia

Un manipulador serial es una **cadena cinemática abierta** de eslabones rígidos conectados por articulaciones. Si posee $n$ articulaciones de un grado de libertad, tiene $n+1$ eslabones, incluida la base fija, que identificamos con el índice $0$.

Numeramos las articulaciones de $1$ a $n$ y los eslabones de $0$ a $n$. La articulación $i$ conecta el eslabón $i-1$ con el eslabón $i$, como ilustra la Figura {numref}`kinematic_chain_num`.

```{figure} img/fkin/kinematic_chain_num.svg
:name: kinematic_chain_num
:align: center
:width: 60%
Numeración de eslabones y articulaciones de una cadena serial.
```

Asociamos un sistema de referencia $\{i\}$ rígidamente al eslabón $i$. Así, las coordenadas de un punto fijo al eslabón permanecen constantes cuando se expresan en su propio sistema, aunque cambien al describirlas desde la base.

```{figure} img/fkin/kinematic_chain_sys.svg
:name: kinematic_chain_sys
:align: center
:width: 60%
Sistemas de referencia asociados a los eslabones.
```

La transformación $T_i^{i-1}$ describe la pose de $\{i\}$ respecto de $\{i-1\}$. Si los sistemas se asignan de manera apropiada a los eslabones y cada articulación posee un grado de libertad, esa transformación depende de la variable $q_i$ correspondiente y de parámetros geométricos constantes.

Por composición, la pose del sistema terminal se obtiene mediante

$$
\boxed{
T_n^0(q)=T_1^0(q_1)\,T_2^1(q_2)\cdots T_n^{n-1}(q_n).
}
$$

El orden de los factores es esencial: las transformaciones homogéneas **no conmutan en general**. Los índices indican qué sistemas se relacionan y ayudan a comprobar que se está utilizando un orden compatible. Si la herramienta requiere un sistema terminal diferente, podremos añadir su transformación correspondiente.

## Diagramas cinemáticos de manipuladores seriales

Un **diagrama cinemático** representa de forma simplificada los eslabones y las articulaciones de un manipulador. Su propósito es mostrar la conectividad y la disposición de los ejes articulares sin reproducir todos los detalles constructivos del robot.

En estos apuntes estudiaremos principalmente manipuladores con articulaciones revolutas (R) y prismáticas (P). La Figura {numref}`kinematic_diagram_symbols` presenta símbolos para representarlas.

```{figure} img/fkin/kinematic_diagram_symbols.svg
:name: kinematic_diagram_symbols
:align: center
:width: 60%
Representación de articulaciones revolutas (R) y prismáticas (P).
```

En una articulación revoluta espacial, el eje del cilindro dibujado indica la **dirección del eje de giro**. En una prismática, la dirección longitudinal del prisma indica la **dirección de deslizamiento**. Los eslabones se representan con segmentos de línea.

El diagrama no pretende describir fielmente la forma o el volumen de cada eslabón. Al realizar el análisis distinguimos las distancias relevantes para la cinemática de los detalles constructivos que no modifican la relación entre las articulaciones.

## Cinemática directa mediante transformaciones elementales

Antes de introducir una convención sistemática, consideremos un manipulador planar **RR** con dos eslabones de longitudes $l_1$ y $l_2$. Ambos ejes de giro son paralelos al eje $z$.

Situamos el sistema $\{0\}$ en la primera articulación, el sistema $\{1\}$ en el extremo del primer eslabón y el sistema $\{2\}$ en el extremo del segundo. Elegimos cada eje $x_i$ en la dirección longitudinal del eslabón correspondiente.

Para pasar de $\{0\}$ a $\{1\}$, rotamos $q_1$ alrededor de $z$ y trasladamos $l_1$ a lo largo del eje $x$ del sistema rotado. Repetimos la construcción para el segundo eslabón:

$$
T_1^0=\operatorname{Rot}_z(q_1)\operatorname{Trans}_x(l_1),
\qquad
T_2^1=\operatorname{Rot}_z(q_2)\operatorname{Trans}_x(l_2).
$$

Por composición,

$$
T_2^0=
\operatorname{Rot}_z(q_1)\operatorname{Trans}_x(l_1)
\operatorname{Rot}_z(q_2)\operatorname{Trans}_x(l_2).
$$

De esta expresión obtenemos las coordenadas del extremo:

$$
x=l_1\cos q_1+l_2\cos(q_1+q_2),
\qquad
y=l_1\sin q_1+l_2\sin(q_1+q_2).
$$

La orientación del sistema terminal en el plano se describe mediante una rotación alrededor de $z$ de ángulo $q_1+q_2$. Más adelante reconstruiremos el modelo completo con Denavit–Hartenberg.

Este ejemplo demuestra que **podemos obtener la cinemática directa sin DH** si conocemos la secuencia de transformaciones elementales. En manipuladores espaciales más complejos, sin embargo, puede resultar difícil elegir sistemas y transformaciones de manera consistente.

### ¿Por qué utilizar una convención sistemática?

La asignación de sistemas de referencia no es única. Diferentes elecciones pueden producir expresiones matriciales distintas, aunque describan el mismo movimiento físico cuando se consideran las transformaciones entre sus respectivas bases y sistemas terminales.

La convención de **Denavit–Hartenberg** proporciona reglas geométricas para asignar sistemas y expresar cada transformación consecutiva mediante cuatro parámetros. Así reducimos decisiones arbitrarias y seguimos un procedimiento común para diferentes robots.

Aquí utilizaremos la **convención DH clásica o estándar**. Existen otras formulaciones, como DH modificada, y otros enfoques, como el producto de exponenciales. Sus reglas de asignación y los órdenes de las transformaciones no deben confundirse.

## Convención de Denavit–Hartenberg

En la convención **DH clásica**, la pose del sistema $\{i\}$ respecto de $\{i-1\}$ se describe mediante cuatro parámetros: $\theta_i$, $d_i$, $a_i$ y $\alpha_i$. La transformación entre sistemas consecutivos se construye en este orden:

$$
\boxed{
T_i^{i-1}=
\operatorname{Rot}_z(\theta_i)
\operatorname{Trans}_z(d_i)
\operatorname{Trans}_x(a_i)
\operatorname{Rot}_x(\alpha_i).
}
$$

Las rotaciones y traslaciones son **transformaciones homogéneas de $4\times4$**, como se definieron en el capítulo anterior. El orden es parte de la convención y no puede alterarse arbitrariamente.

Al efectuar el producto obtenemos

$$
T_i^{i-1}=
\begin{bmatrix}
c_{\theta_i}&-s_{\theta_i}c_{\alpha_i}&s_{\theta_i}s_{\alpha_i}&a_ic_{\theta_i}\\
s_{\theta_i}&c_{\theta_i}c_{\alpha_i}&-c_{\theta_i}s_{\alpha_i}&a_is_{\theta_i}\\
0&s_{\alpha_i}&c_{\alpha_i}&d_i\\
0&0&0&1
\end{bmatrix}.
$$ (dh_matrix)

Aquí $c_\varphi=\cos\varphi$ y $s_\varphi=\sin\varphi$. Los parámetros describen la geometría relativa de los sistemas; uno de ellos incorpora la variable articular de la junta $i$.

### Asignación de los sistemas de referencia

La matriz {eq}`dh_matrix` tiene la forma anterior **solo cuando los sistemas se asignan de acuerdo con la convención DH clásica**. Comenzamos por identificar los ejes de las articulaciones.

**1. Ejes $z$.** El eje $z_{i-1}$ se coloca sobre el eje de movimiento de la articulación $i$: eje de giro para una revoluta o dirección de deslizamiento para una prismática, con $i=1,\ldots,n$.

**2. Sistema base $\{0\}$.** Su origen se sitúa sobre el eje $z_0$ de la primera articulación. Los ejes $x_0$ e $y_0$ se eligen para completar un sistema derecho. La elección del sistema base influye en los parámetros y en la pose calculada respecto de él.

**3. Sistemas intermedios.** Para definir el sistema $\{i\}$, examinamos las rectas que contienen $z_{i-1}$ y $z_i$. En la convención DH clásica, el eje $x_i$ debe cumplir **dos condiciones geométricas respecto de $z_{i-1}$**:

1. **Ser perpendicular a $z_{i-1}$.**
2. **Intersecar a $z_{i-1}$.**

Estas condiciones se refieren a las rectas que contienen los ejes, no a que sus orígenes deban coincidir. Para completar la construcción, $x_i$ se elige como una **normal común** a $z_{i-1}$ y $z_i$: también es perpendicular a $z_i$ y lo interseca en el origen $o_i$ del sistema $\{i\}$. De esta manera, la elección de $x_i$ es compatible con la secuencia de transformaciones de la matriz DH clásica.

La manera de trazar esa normal común depende de la disposición relativa de los ejes $z$:

- **Ejes no coplanares (rectas alabeadas):** los ejes no son paralelos ni se intersecan. Existe una única normal común que une ambas rectas con distancia mínima; su dirección define $x_i$.
- **Ejes paralelos y distintos:** existen múltiples normales comunes. Una elección habitual es la que pasa por el origen anterior.
- **Ejes que se intersecan:** $x_i$ se elige perpendicular al plano de ambos ejes, con dirección compatible con $z_{i-1}\times z_i$.
- **Ejes coincidentes:** puede elegirse $x_i$ perpendicular al eje común y alineado con $x_{i-1}$ en una configuración de referencia. Esto no implica que la alineación se mantenga durante el movimiento.

En todos los casos, $y_i$ se determina para completar un sistema coordenado derecho.

**4. Sistema terminal $\{n\}$.** Una vez identificado el eje $z_{n-1}$ de la última articulación, el sistema final admite cierta libertad de elección. Es frecuente tomar $z_n$ paralelo a $z_{n-1}$ para simplificar el modelo, pero **no es una obligación universal**. El último sistema DH tampoco tiene que coincidir con el punto de operación de la herramienta.

Estas reglas permiten comprobar la disposición geométrica de los sistemas antes de obtener los parámetros. En particular, al revisar un diagrama DH, conviene verificar que cada eje $x_i$ sea perpendicular a $z_{i-1}$ y lo interseque.

### Parámetros de Denavit–Hartenberg

Una vez establecidos los sistemas, obtenemos los parámetros DH. Antes de identificarlos en un manipulador conviene visualizar **dónde se miden**. En la Figura {numref}`dh_parameters` se representa un caso general de dos ejes consecutivos no coplanares. El eje $x_i$ es una normal común que corta a $z_{i-1}$ en el punto $H$ y a $z_i$ en el origen $o_i$.

```{figure} img/fkin/dh_parameters.svg
:name: dh_parameters
:align: center
:width: 50%
Interpretación geométrica de los cuatro parámetros de Denavit–Hartenberg clásicos: $d_i$ y $\theta_i$ se miden a lo largo y alrededor de $z_{i-1}$, respectivamente; $a_i$ y $\alpha_i$ se miden a lo largo y alrededor de $x_i$, respectivamente. Los ejes punteados representan direcciones trasladadas paralelamente para comparar orientaciones.
```

Los parámetros son desplazamientos y ángulos **con signo**, medidos según el sentido positivo de los ejes elegidos:

- **$\theta_i$:** ángulo de rotación alrededor de $z_{i-1}$ desde la dirección de $x_{i-1}$ hacia la de $x_i$. Para comparar ambas direcciones, puede trasladarse paralelamente $x_{i-1}$ hasta el punto $H$.
- **$d_i$:** desplazamiento a lo largo de $z_{i-1}$ desde el origen $o_{i-1}$ hasta el punto $H$, donde el eje $x_i$ interseca a $z_{i-1}$.
- **$a_i$:** desplazamiento a lo largo de $x_i$ desde $H$ hasta el origen $o_i$, donde $x_i$ interseca a $z_i$.
- **$\alpha_i$:** ángulo de rotación alrededor de $x_i$ desde la dirección de $z_{i-1}$ hasta la de $z_i$. Para interpretarlo, puede trasladarse paralelamente $z_{i-1}$ hasta el origen $o_i$.

En resumen, **$\theta_i$ y $d_i$ se asocian al eje $z_{i-1}$**, mientras que **$a_i$ y $\alpha_i$ se asocian al eje $x_i$**. Esta agrupación permite recordar el orden de operaciones de la transformación $T_i^{i-1}$. Aunque la figura muestra ejes no coplanares para distinguir las cuatro mediciones, las mismas definiciones se aplican cuando los ejes son paralelos, se intersecan o coinciden, utilizando las reglas de asignación explicadas anteriormente.

En una articulación **R**, el parámetro variable es $\theta_i$ y $d_i$ es constante. En una articulación **P**, el parámetro variable es $d_i$ y $\theta_i$ es constante. Los parámetros geométricos $a_i$ y $\alpha_i$ son constantes en ambos casos.

Conviene distinguir el parámetro DH de la coordenada articular empleada por el controlador o el enunciado. Según el origen y el sentido elegidos para $q_i$, puede ser necesario incorporar un desplazamiento constante o invertir el signo. Si los sentidos positivos coinciden, por ejemplo:

$$
\theta_i=q_i+\theta_{i,0}\quad\text{(revoluta)},
\qquad
d_i=q_i+d_{i,0}\quad\text{(prismática)}.
$$

No debemos suponer que el cero articular del robot físico coincide automáticamente con el utilizado para construir su tabla DH.

Una vez determinados los parámetros de todos los sistemas, los organizamos como se muestra en la Tabla {numref}`dh_table`. Cada fila representa una articulación y reúne la información para construir su transformación $T_i^{i-1}$.

```{table} Parámetros de Denavit–Hartenberg (convención clásica).
:name: dh_table

| $i$ | $a_i$ | $\alpha_i$ | $d_i$ | $\theta_i$ |
|:---:|:-----:|:----------:|:-----:|:----------:|
| $1$ | $a_1$ | $\alpha_1$ | $d_1$ | $\theta_1$ |
| $2$ | $a_2$ | $\alpha_2$ | $d_2$ | $\theta_2$ |
| $\vdots$ | $\vdots$ | $\vdots$ | $\vdots$ | $\vdots$ |
| $n$ | $a_n$ | $\alpha_n$ | $d_n$ | $\theta_n$ |

```

La tabla contiene una fila por articulación de un grado de libertad. No describe necesariamente transformaciones adicionales entre la base DH y el sistema del mundo, ni entre el último sistema DH y el punto de operación de una herramienta.

### Construcción del modelo de cinemática directa

Con los sistemas de referencia asignados y una tabla de parámetros como la Tabla {numref}`dh_table`, sustituimos cada fila en la ecuación {eq}`dh_matrix` para obtener las transformaciones consecutivas $T_i^{i-1}$.

La pose del sistema terminal respecto de la base es el producto ordenado

$$
T_n^0(q)=T_1^0(q_1)T_2^1(q_2)\cdots T_n^{n-1}(q_n).
$$

Para interpretar el resultado separamos la orientación y la posición:

$$
T_n^0(q)=
\begin{bmatrix}
R_n^0(q)&\vec r_{o_n}^{\,0}(q)\\
\vec 0^{\,T}&1
\end{bmatrix}.
$$

La matriz depende de las coordenadas articulares y de los parámetros geométricos. A continuación aplicaremos este procedimiento a manipuladores sencillos y después a un robot industrial de seis articulaciones.


**Solución.** El robot tiene dos articulaciones revolutas de ejes paralelos y eslabones de longitudes $l_1$ y $l_2$. Colocamos $z_0$ sobre el eje de la primera articulación y $z_1$ sobre el de la segunda. Tomamos $z_2$ paralelo a los anteriores en el extremo del segundo eslabón. Los ejes $x_1$ y $x_2$ apuntan a lo largo de sus eslabones respectivos, como se muestra en la parte b de la Figura {numref}`fk_rr_robot_dh`.

Como los ejes $z$ son paralelos y tienen el mismo sentido, los ángulos de torsión $\alpha_1$ y $\alpha_2$ son nulos. Las normales comunes tienen longitudes $l_1$ y $l_2$, de modo que $a_1=l_1$ y $a_2=l_2$. Los orígenes elegidos están a la misma altura, por lo cual $d_1=d_2=0$. Finalmente, al ser articulaciones revolutas, $\theta_1=q_1$ y $\theta_2=q_2$.

| $i$ | $a_i$ | $\alpha_i$ | $d_i$ | $\theta_i$ |
|:---:|:---:|:---:|:---:|:---:|
| $1$ | $l_1$ | $0$ | $0$ | $q_1$ |
| $2$ | $l_2$ | $0$ | $0$ | $q_2$ |

De la ecuación {eq}`dh_matrix` obtenemos:

$$
T_1^0 =
\begin{bmatrix}
c_1&-s_1&0&l_1c_1\\
s_1&c_1&0&l_1s_1\\
0&0&1&0\\
0&0&0&1
\end{bmatrix},
\qquad
T_2^1 =
\begin{bmatrix}
c_2&-s_2&0&l_2c_2\\
s_2&c_2&0&l_2s_2\\
0&0&1&0\\
0&0&0&1
\end{bmatrix}.
$$

Aquí $c_i=\cos q_i$ y $s_i=\sin q_i$. Tras multiplicar las transformaciones y utilizar las identidades trigonométricas

$$
c_{12}=\cos(q_1+q_2),\qquad s_{12}=\sin(q_1+q_2),
$$

obtenemos el modelo:

$$
\boxed{
T_2^0=T_1^0T_2^1=
\begin{bmatrix}
c_{12}&-s_{12}&0&l_1c_1+l_2c_{12}\\
s_{12}&c_{12}&0&l_1s_1+l_2s_{12}\\
0&0&1&0\\
0&0&0&1
\end{bmatrix}.
}
$$

La cuarta columna proporciona la **posición del extremo** en la base, mientras que el bloque superior izquierdo representa una rotación de $q_1+q_2$ alrededor de $z$. Este resultado coincide con el obtenido previamente mediante transformaciones elementales.

**Comprobación.** Para $q_1=q_2=0$, los dos eslabones quedan alineados sobre el eje $x_0$ y el extremo debe estar en $(l_1+l_2,0,0)$. Al sustituir ambos ángulos en $T_2^0$ obtenemos exactamente esa posición y una orientación igual a la de la base.

:::

### Comprobación simbólica del manipulador RR con SymPy

Una vez obtenido el modelo de manera analítica, podemos utilizar **SymPy** para reproducir el producto de matrices y comprobar las identidades trigonométricas. El siguiente programa construye la matriz DH clásica a partir de sus cuatro parámetros, calcula $T_2^0$ y **muestra la matriz obtenida** para compararla con la expresión deducida anteriormente. Finalmente, añade una comprobación automática de que ambas coinciden.

El código constituye una **comprobación**, no un sustituto de la asignación geométrica de sistemas ni de la determinación de los parámetros. Puedes ejecutarlo en un notebook de Python para visualizar la matriz simbólica; la construcción del libro no ejecuta automáticamente estos fragmentos.

```python
import sympy as sp

q1, q2, l1, l2 = sp.symbols("q1 q2 l1 l2", real=True)

def dh(a, alpha, d, theta):
    ct, st = sp.cos(theta), sp.sin(theta)
    ca, sa = sp.cos(alpha), sp.sin(alpha)
    return sp.Matrix([
        [ct, -st*ca,  st*sa, a*ct],
        [st,  ct*ca, -ct*sa, a*st],
        [0,       sa,     ca,    d],
        [0,        0,      0,    1],
    ])

T = sp.trigsimp(dh(l1, 0, 0, q1) * dh(l2, 0, 0, q2))
print("Transformación T_2^0 calculada con SymPy:")
sp.pprint(T)

T_esperada = sp.Matrix([
    [sp.cos(q1 + q2), -sp.sin(q1 + q2), 0,
     l1*sp.cos(q1) + l2*sp.cos(q1 + q2)],
    [sp.sin(q1 + q2),  sp.cos(q1 + q2), 0,
     l1*sp.sin(q1) + l2*sp.sin(q1 + q2)],
    [0, 0, 1, 0],
    [0, 0, 0, 1],
])

assert sp.simplify(T - T_esperada) == sp.zeros(4)
print("El modelo simbólico coincide con la solución analítica.")
```

Primero, **compara visualmente** la matriz mostrada con la solución analítica del ejemplo RR. La instrucción `assert` añade una comprobación automática: si la diferencia simplificada entre ambas matrices es nula, el programa continúa e imprime el mensaje de confirmación; si no lo es, señala que la comparación no se cumplió. La validación es simbólica y no requiere asignar valores numéricos a $q_1$, $q_2$, $l_1$ o $l_2$.

Los ejes $z_0$ y $z_1$ se intersecan; de ahí que $a_1=0$ y, con la orientación mostrada, $\alpha_1=90^\circ$. Los ejes $z_1$, $z_2$ y $z_3$ son paralelos, por lo que $\alpha_2=\alpha_3=0$. Las distancias de los eslabones corresponden a $a_2=l_2$ y $a_3=l_3$. El origen del segundo eje se encuentra a una altura constante $d_1$ respecto de la base.

| $i$ | $a_i$ | $\alpha_i$ | $d_i$ | $\theta_i$ |
|:---:|:---:|:---:|:---:|:---:|
| $1$ | $0$ | $90^\circ$ | $d_1$ | $q_1$ |
| $2$ | $l_2$ | $0$ | $0$ | $q_2$ |
| $3$ | $l_3$ | $0$ | $0$ | $q_3$ |

Al sustituir en la matriz DH, obtenemos

$$
T_1^0=
\begin{bmatrix}
c_1&0&s_1&0\\
s_1&0&-c_1&0\\
0&1&0&d_1\\
0&0&0&1
\end{bmatrix},
\qquad
T_2^1=
\begin{bmatrix}
c_2&-s_2&0&l_2c_2\\
s_2&c_2&0&l_2s_2\\
0&0&1&0\\
0&0&0&1
\end{bmatrix},
$$

$$
T_3^2=
\begin{bmatrix}
c_3&-s_3&0&l_3c_3\\
s_3&c_3&0&l_3s_3\\
0&0&1&0\\
0&0&0&1
\end{bmatrix}.
$$

Es conveniente multiplicar primero las dos últimas matrices, cuyos ejes de giro son paralelos. Si definimos $c_{23}=\cos(q_2+q_3)$ y $s_{23}=\sin(q_2+q_3)$, el producto se simplifica a

$$
T_2^1T_3^2=
\begin{bmatrix}
c_{23}&-s_{23}&0&l_2c_2+l_3c_{23}\\
s_{23}&c_{23}&0&l_2s_2+l_3s_{23}\\
0&0&1&0\\
0&0&0&1
\end{bmatrix}.
$$

Finalmente,

$$
\boxed{
T_3^0=
\begin{bmatrix}
c_1c_{23}&-c_1s_{23}&s_1&c_1(l_2c_2+l_3c_{23})\\
s_1c_{23}&-s_1s_{23}&-c_1&s_1(l_2c_2+l_3c_{23})\\
s_{23}&c_{23}&0&d_1+l_2s_2+l_3s_{23}\\
0&0&0&1
\end{bmatrix}.
}
$$

**Interpretación.** La distancia horizontal al eje de la primera articulación es $l_2c_2+l_3c_{23}$; el ángulo $q_1$ determina su dirección en el plano $xy$. La coordenada vertical es $d_1+l_2s_2+l_3s_{23}$. Cuando $q_2=q_3=0$, el extremo está a altura $d_1$ y a distancia radial $l_2+l_3$, lo que permite comprobar los términos de posición del modelo.

Aunque el manipulador tiene tres articulaciones, no puede alcanzar arbitrariamente todas las orientaciones espaciales: sus movimientos y su arquitectura restringen las poses posibles.

**Ejemplo. Cinemática directa de un manipulador esférico RRP.** Consideremos un manipulador con dos articulaciones revolutas cuyos ejes son perpendiculares y se intersecan en un punto $O$. La tercera articulación es prismática y desplaza el extremo en dirección radial. Determine su transformación homogénea respecto de la base.

```{figure} img/fkin/rrp_spherical.svg
:name: fk_rrp_spherical
:align: center
:width: 85%
Esquema de un manipulador esférico RRP. Los dos primeros ejes articulares se intersecan en $O$; la articulación prismática se desplaza a lo largo de la dirección radial.
```

**Solución.** Tomamos el origen del sistema base $\{0\}$ en $O$ y su eje $z_0$ vertical. El eje $z_1$ corresponde a la segunda articulación revoluta y es perpendicular a $z_0$. El eje $z_2$ sigue la dirección de desplazamiento de la articulación prismática; por ello, al cambiar $q_2$, cambia también la dirección radial.

Elegimos los sistemas $\{1\}$ y $\{2\}$ con origen en $O$, y el sistema terminal $\{3\}$ con origen en el extremo del elemento prismático. Los ejes $x_1$ y $x_2$ se colocan sobre las normales comunes a las parejas de ejes $z_0,z_1$ y $z_1,z_2$, respectivamente. Como ambas parejas de ejes se intersecan, sus distancias normales son nulas: $a_1=a_2=0$. El ángulo de torsión para la primera pareja se toma como $\alpha_1=-90^\circ$ y para la segunda como $\alpha_2=90^\circ$, según el sentido de los ejes elegidos.

Definimos $q_1$ como el **ángulo azimutal** alrededor del eje vertical, $q_2$ como el **ángulo de inclinación** de la dirección radial respecto de $+z_0$ y $q_3$ como la **distancia radial** desde $O$ hasta el extremo. Elegimos los ceros articulares de manera que, cuando $q_1=q_2=0$, la dirección de la articulación prismática coincida con $+z_0$. Para esta elección, la tabla DH clásica es:

| $i$ | $a_i$ | $\alpha_i$ | $d_i$ | $\theta_i$ |
|:---:|:---:|:---:|:---:|:---:|
| $1$ | $0$ | $-90^\circ$ | $0$ | $q_1$ |
| $2$ | $0$ | $90^\circ$ | $0$ | $q_2$ |
| $3$ | $0$ | $0$ | $q_3$ | $0$ |

En la tercera fila, **$d_3=q_3$ es el parámetro variable**, mientras que $\theta_3=0$ permanece constante. Las transformaciones se escriben de manera compacta como

$$
T_1^0=\operatorname{Rot}_z(q_1)\operatorname{Rot}_x(-90^\circ),\qquad
T_2^1=\operatorname{Rot}_z(q_2)\operatorname{Rot}_x(90^\circ),\qquad
T_3^2=\operatorname{Trans}_z(q_3).
$$

Al multiplicarlas y escribir $c_i=\cos q_i$ y $s_i=\sin q_i$, resulta

$$
\boxed{
T_3^0=
\begin{bmatrix}
c_1c_2 & -s_1 & c_1s_2 & q_3c_1s_2\\
s_1c_2 & c_1 & s_1s_2 & q_3s_1s_2\\
-s_2 & 0 & c_2 & q_3c_2\\
0 & 0 & 0 & 1
\end{bmatrix}.
}
$$

La posición del extremo está dada por

$$
x=q_3\cos q_1\sin q_2,\qquad
y=q_3\sin q_1\sin q_2,\qquad
z=q_3\cos q_2.
$$

Son precisamente las expresiones de las **coordenadas esféricas** para una distancia radial $q_3$, un ángulo azimutal $q_1$ y una inclinación $q_2$ medida desde el eje vertical positivo. Esto ilustra la relación entre la morfología del robot y la geometría de su espacio de trabajo.

**Comprobación.** Si $q_2=0$, la dirección radial coincide con $+z_0$ y la posición es $(0,0,q_3)$, independientemente del valor de $q_1$. Si $q_1=0$ y $q_2=90^\circ$, el extremo queda en $(q_3,0,0)$. En ambos casos, el resultado es coherente con la geometría del robot.

El modelo es idealizado: supone ejes revolutos que se intersecan, desplazamiento radial medido desde $O$ y ausencia de longitudes u offsets adicionales. En un manipulador real podrían incorporarse estos parámetros sin modificar el procedimiento DH.

:::

---

:::{div}
:class: example

**Ejemplo. Cinemática directa de un robot industrial ABB IRB 140.** Construyamos un modelo geométrico de seis articulaciones revolutas a partir de las dimensiones y los ejes articulares del manipulador.

```{figure} img/fkin/abb-irb-140-dimensions.png
:name: abb-irb-140-dimensions
:align: center
:width: 60%
Dimensiones del robot ABB IRB 140 utilizadas en el modelo.
```

### Evaluación numérica de un robot industrial con NumPy

Para un manipulador de seis articulaciones, resulta más práctico evaluar las transformaciones numéricamente que desarrollar todas las entradas del producto simbólico. En el siguiente ejemplo utilizamos **NumPy** y los parámetros DH del ABB IRB 140 definidos anteriormente.

La tabla del ejemplo expresa los ángulos en grados, mientras que las funciones trigonométricas de NumPy utilizan **radianes**. Por eso realizamos la conversión explícitamente. Conservamos las distancias en milímetros y multiplicamos las transformaciones en el mismo orden que la cadena cinemática.

```python
import numpy as np

def dh(a, alpha, d, theta):
    ct, st = np.cos(theta), np.sin(theta)
    ca, sa = np.cos(alpha), np.sin(alpha)
    return np.array([
        [ct, -st*ca,  st*sa, a*ct],
        [st,  ct*ca, -ct*sa, a*st],
        [0,       sa,     ca,    d],
        [0,        0,      0,    1],
    ], dtype=float)

# (a_i en mm, alpha_i en grados, d_i en mm)
parametros = [
    (70, 90, 352),
    (360, 0, 0),
    (0, 90, 0),
    (0, -90, 380),
    (0, 90, 0),
    (0, 0, 65),
]
q = np.deg2rad([0, 90, 0, 0, 0, 0])

T = np.eye(4)
for (a, alpha_grados, d), theta in zip(parametros, q):
    T = T @ dh(a, np.deg2rad(alpha_grados), d, theta)

np.set_printoptions(precision=3, suppress=True)
print("Transformación T_6^0 calculada con NumPy (posición en mm):")
print(T)

T_esperada = np.array([
    [0, 0, 1, 515],
    [0, -1, 0, 0],
    [1, 0, 0, 712],
    [0, 0, 0, 1],
])
np.testing.assert_allclose(T, T_esperada, atol=1e-10)
np.testing.assert_allclose(T[:3, :3].T @ T[:3, :3], np.eye(3), atol=1e-10)
np.testing.assert_allclose(np.linalg.det(T[:3, :3]), 1.0, atol=1e-10)
print("Las comprobaciones numéricas se cumplieron.")
```

Al ejecutar el código, la matriz impresa permite **comparar visualmente** el resultado con el obtenido en el ejemplo: la posición del extremo es $(515,0,712)$ mm. El resultado esperado, omitiendo residuos numéricos cercanos a cero, es

```text
[[  0.   0.   1. 515.]
 [  0.  -1.   0.   0.]
 [  1.   0.   0. 712.]
 [  0.   0.   0.   1.]]
```

Algunas entradas podrían aparecer como `-0.` por efectos del redondeo en punto flotante; se interpretan como cero. Las instrucciones `np.testing.assert_allclose` realizan además comprobaciones automáticas con una tolerancia pequeña, necesaria por el redondeo numérico. Las dos últimas verifican que el bloque de orientación es ortogonal y tiene determinante unitario. Si alguna comparación falla, el programa indica el error; si todas se cumplen, aparece el mensaje de confirmación.

Este programa utiliza exclusivamente **NumPy** y no depende de una biblioteca especializada de robótica. Más adelante podremos utilizar herramientas de mayor nivel para modelar y visualizar manipuladores sin repetir todo el cálculo matricial.

## Interpretación y comprobación de resultados

Obtener el producto de matrices es solo una parte del análisis. También necesitamos comprobar que el modelo representa la geometría y las convenciones que hemos adoptado.

Una primera comprobación consiste en **evaluar una configuración conocida** —por ejemplo, con algunas articulaciones en cero— y comparar la pose calculada con la que se espera del diagrama cinemático. Las longitudes y desplazamientos deben expresarse en unidades compatibles.

Si extraemos la submatriz de orientación $R_n^0$ de una transformación homogénea válida, debe cumplirse

$$
(R_n^0)^T R_n^0=I_3,
\qquad
\det(R_n^0)=1.
$$

La última fila de la transformación debe ser $[0\;0\;0\;1]$. En cálculos numéricos pueden aparecer pequeñas diferencias por redondeo; las discrepancias considerables suelen indicar errores en el orden de composición, los signos de los ángulos o la tabla de parámetros.

El sistema $\{n\}$ no tiene por qué coincidir con el **punto central de herramienta** (*Tool Center Point*, TCP). Si conocemos una transformación fija desde el último sistema DH hasta la herramienta, podemos obtener

$$
T_{\mathrm{TCP}}^0(q)=T_n^0(q)\,T_{\mathrm{TCP}}^n.
$$

En aplicaciones computacionales posteriores podremos utilizar Python para evaluar estas expresiones y contrastar los resultados simbólicos, numéricos y geométricos.

### Guía para construir y revisar un modelo DH

En un problema nuevo, conviene seguir una secuencia de trabajo definida:

1. **Identificar** las articulaciones y numerarlas desde la base, señalando sus ejes de movimiento y variables $q_i$.
2. **Asignar** los sistemas de referencia según las reglas de DH clásica, comprobando las condiciones geométricas de los ejes $x_i$ y $z_{i-1}$.
3. **Obtener** los cuatro parámetros de cada articulación e indicar los desplazamientos constantes, los sentidos positivos y las unidades.
4. **Construir** cada transformación $T_i^{i-1}$ y multiplicarlas en el orden de la cadena para obtener $T_n^0(q)$.
5. **Interpretar y comprobar** la posición y orientación resultantes en al menos una configuración cuya geometría se comprenda fácilmente.

Este procedimiento no determina una única tabla DH para un robot: la asignación de sistemas y los ceros articulares pueden variar. Lo importante es que la tabla, las transformaciones y el significado de las variables sean coherentes entre sí.
