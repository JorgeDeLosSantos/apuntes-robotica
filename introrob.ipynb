# Una introducción a los manipuladores seriales

## ¿Qué es un robot?

De acuerdo con la norma [ISO 8373:2021](https://www.iso.org/standard/75539.html), un **robot** es un mecanismo accionado y programado que posee cierto grado de **autonomía** para realizar tareas de locomoción, manipulación o posicionamiento. La autonomía es la capacidad de ejecutar tareas previstas a partir del estado del sistema y de la información que percibe, sin requerir intervención humana en cada decisión.

Esta definición es más amplia que la de un brazo robótico: comprende, entre otros, robots manipuladores y robots móviles. Un mecanismo desarrollado con tecnología robótica que no satisfaga todas las características de un robot —por ejemplo, un manipulador completamente teleoperado— puede denominarse **dispositivo robótico**.

Según su finalidad, la terminología de ISO 8373:2021 distingue las siguientes categorías:

- **Robot industrial:** manipulador multipropósito, reprogramable y controlado automáticamente, programable en tres o más ejes, que puede estar instalado en una posición fija o sobre una plataforma móvil para aplicaciones de automatización en entornos industriales.
- **Robot de servicio:** robot que realiza tareas útiles para personas o equipos, en aplicaciones personales o profesionales.
- **Robot médico:** robot destinado a utilizarse como equipo o sistema electromédico.

Estas categorías se refieren principalmente al uso y a los requisitos del sistema, no a una configuración mecánica exclusiva. Dos robots con estructuras semejantes pueden utilizarse en aplicaciones distintas. Para ampliar las definiciones, pueden consultarse las páginas de la *International Federation of Robotics* (IFR) sobre [robots industriales](https://ifr.org/industrial-robots) y [robots de servicio](https://ifr.org/service-robots).

En particular, la definición de robot industrial incluye tres características:

- **Control automático:** el sistema de control ejecuta movimientos y funciones auxiliares de acuerdo con un programa de tareas.
- **Reprogramabilidad:** los movimientos y las funciones pueden modificarse mediante programación, sin cambiar físicamente el mecanismo.
- **Carácter multipropósito:** el sistema puede adaptarse a aplicaciones diferentes, incluso si ello requiere modificar herramientas u otros elementos físicos.

En estos apuntes nos concentraremos en el **análisis cinemático y dinámico de manipuladores seriales**, especialmente los utilizados en robótica industrial. Por ello, cuando el contexto sea inequívoco, podremos referirnos a ellos como *robots seriales* o *brazos robóticos*. No obstante, **robot**, **robot industrial** y **manipulador serial** no son términos equivalentes.


## Manipuladores robóticos y cadenas cinemáticas

Un **manipulador** es un mecanismo constituido por elementos conectados mediante articulaciones que permiten producir movimientos relativos entre ellos. En robótica, estos mecanismos se emplean para posicionar y orientar herramientas, piezas u otros dispositivos. El manipulador es la parte mecánica del sistema: el robot también comprende elementos como los actuadores, el controlador y los medios de programación.

Para describir un manipulador identificamos cuatro componentes fundamentales:

- **Base:** elemento desde el cual se conecta o sostiene la estructura mecánica del manipulador.
- **Eslabones:** cuerpos que forman la estructura mecánica y conectan unas articulaciones con otras.
- **Articulaciones o juntas:** uniones que permiten determinados movimientos relativos entre eslabones.
- **Efector final:** dispositivo instalado en el extremo del manipulador para interactuar con su entorno; puede ser una pinza, una herramienta de soldadura u otro elemento de trabajo.

Una **cadena cinemática** es un conjunto de eslabones conectados por articulaciones. En un **manipulador serial**, los eslabones forman una cadena abierta: existe una única trayectoria de conexiones entre la base y el extremo de la cadena. En un **manipulador paralelo**, la plataforma móvil se conecta con la base mediante dos o más cadenas cinemáticas, formando lazos cerrados. También existen arquitecturas híbridas que combinan ambas disposiciones.

### Articulaciones y grados de libertad

Entre las articulaciones más utilizadas en manipuladores seriales se encuentran:

- **Revoluta (R):** permite una rotación relativa alrededor de un eje. Su variable articular suele representarse mediante un ángulo, como $\theta_i$.
- **Prismática (P):** permite una traslación relativa a lo largo de un eje. Su variable articular suele representarse mediante un desplazamiento, como $d_i$.

Los **grados de libertad** indican la cantidad de coordenadas independientes necesarias para describir la configuración de un mecanismo. Por ejemplo, un manipulador plano formado por dos articulaciones revolutas independientes (2R) tiene dos grados de libertad, y su configuración puede expresarse mediante el vector $\mathbf{q} = [\theta_1,\ \theta_2]^\mathsf{T}$.

El estudio de estos mecanismos da lugar a dos problemas centrales: la **cinemática**, que describe las posiciones, orientaciones y movimientos sin considerar sus causas, y la **dinámica**, que relaciona el movimiento con las fuerzas y los pares que lo producen. Estos serán los temas principales de los capítulos siguientes.


## Clasificación de robots de acuerdo con su estructura mecánica

De acuerdo con la IFR (International Federation of Robotics) los robots se pueden clasificar por su estructura mecánica de la siguiente manera:

* Robot articulado 
* Robot cartesiano
* Robot cilíndrico
* Robot paralelo
* Robot SCARA
* Otros

Un *robot articulado* es aquel cuyas articulaciones son todas revolutas. La mayoría de los manipuladores industriales tienen esta estructura mecánica. Se suele denominar a este robot también como antropomórfico, por la similitud de su estructura con los movimientos del cuerpo humano. Un *robot paralelo* es aquel en que posee dos o más cadenas cinemáticas que conectan la base (plataforma fija) con el efector final (plataforma móvil). La cinemática de cadena cerrada de los robots paralelos puede dar como resultado una mayor rigidez estructural y, por lo tanto, una mayor precisión que los robots de cadena cinemática abierta. En la Figura {numref}`abb_irb_360` se puede observar el robot paralelo ABB IRB 360 FlexPicker, el cual de acuerdo con el fabricante es capaz de las aplicaciones de recolección más rápidas y se ha optimizado para aplicaciones de embalaje.

## Algunos datos sobre los robots industriales

En esta sección se listan algunos datos reportados en el World Robotics 2021 de la IFR. En 2020, el stock operativo de robots industriales se calcula en $3,014,879$ unidades ($+10\%$). En la Figura {numref}`operational_stock` puedes observar el crecimiento de robots industriales instalados en fábricas de todo el mundo.

La IFR reporta que Asia es el mercado de robots industriales más grande del mundo. En 2020, se instalaron $266,452$ unidades, un $7\%$ más que las $249,598$ unidades de 2019. Las instalaciones en China crecieron con fuerza ($168,377$ unidades; $+20\%$), mientras que el mercado japonés ($38,653$ unidades; $-23\%$) y el mercado coreano ($30,506$ unidades; $-7\%$) tuvieron ciertas dificultades. Las instalaciones de robots en el segundo mercado más grande, Europa, disminuyeron un 8\% a $67,700$unidades, por segundo año consecutivo desde un máximo de $75,560$ unidades en $2018$. En el continente Americano, las instalaciones cayeron un $17\%$ a $38,736$ unidades en $2020$. Al mayor mercado en esta región, Estados Unidos, le ha ido mejor ($30,787$ unidades; $-8\%$ en 2020) que los otros mercados importantes de México ($3,363$ unidades; $-26\%$) y Canadá ($2,566$ unidades; $-29\%$).
